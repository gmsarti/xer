# Análise das Tabelas e Classificações

Este notebook demonstra, usando **pandas**, como as tabelas foram criadas no banco `contos.sqlite` e quais classificações já foram inseridas.

```python
import pandas as pd
import sqlite3

# caminho para o banco de dados
db_path = '../data/contos.sqlite'
conn = sqlite3.connect(db_path)
```

## 1. Listagem das tabelas existentes

```python
tables = pd.read_sql_query("SELECT name FROM sqlite_master WHERE type='table'", conn)
display(tables)
```

## 2. Esquema das tabelas (PRAGMA table_info)

```python
def show_schema(table_name):
    query = f'PRAGMA table_info({{table_name}})'
    df = pd.read_sql_query(query, conn)
    df.columns = ['cid', 'name', 'type', 'notnull', 'dflt_value', 'pk']
    display(df)

for t in tables['name']:
    print(f'### Schema da tabela: {{t}}')
    show_schema(t)
    print('')
```

## 3. Dados de classificação inseridos

```python
classificacoes = pd.read_sql_query('SELECT * FROM classificacoes', conn)
display(classificacoes)
```

## 4. Relação entre contos e classificações (tabela de ligação)

```python
query = '''
SELECT t.id AS conto_id, t.titulo, c.nome_classificacao, c.framework
FROM tales t
JOIN conto_classificacao lc ON t.id = lc.conto_id
JOIN classificacoes c ON lc.classificacao_id = c.classificacao_id
ORDER BY t.id
LIMIT 20
'''
df = pd.read_sql_query(query, conn)
display(df)
```

## 5. Conclusão

- As tabelas `classificacoes` e `conto_classificacao` foram criadas para armazenar classificações genéricas (ATU, Booker, papéis de Propp).
- O script de classificação já inseriu várias linhas de exemplo, que podem ser visualizadas na última consulta.
