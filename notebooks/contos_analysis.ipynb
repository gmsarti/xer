# Análise dos Contos

Este notebook demonstra como conectar ao banco de dados SQLite `contos.db`, carregar as tabelas em `pandas` e fazer análises exploratórias básicas.

> **Nota:** Substitua os nomes de tabelas/colunas conforme o seu esquema.

```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
sns.set(style="whitegrid")
```

```python
# Caminho para o banco de dados (assumindo que está na pasta data)
db_path = '../data/contos.sqlite'
conn = sqlite3.connect(db_path)
```

```python
# Listar as tabelas disponíveis
tables_query = "SELECT name FROM sqlite_master WHERE type='table';"
tables = pd.read_sql_query(tables_query, conn)
print('Tabelas encontradas:')
print(tables['name'].tolist())
```

```python
# Carregar a tabela principal (substitua 'contos' pelo nome real)
table_name = 'classificacoes'  # <-- ajuste se necessário
df = pd.read_sql_query(f'SELECT * FROM {table_name}', conn)
```

```python
# Visualizar as primeiras linhas
df.sample(10)
```

```python
# Estatísticas descritivas básicas
df.describe(include='all')
```

```python
# Exemplo de visualização: distribuição do tamanho dos contos (supondo coluna 'texto')
if 'texto' in df.columns:
    df['tamanho'] = df['texto'].str.len()
    plt.figure(figsize=(8,5))
    sns.histplot(df['tamanho'], bins=30, kde=True)
    plt.title('Distribuição do tamanho dos contos')
    plt.xlabel('Número de caracteres')
    plt.show()
else:
    print('Coluna "texto" não encontrada. Ajuste o nome da coluna conforme seu esquema.')
```
